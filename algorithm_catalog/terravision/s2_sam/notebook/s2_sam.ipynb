```python
import openeo

connection = openeo.connect("openeofed.dataspace.copernicus.eu").authenticate_oidc()
```

```python
spatial_extent = {
  "type": "FeatureCollection",
  "name": "slovenia",
  "crs": { "type": "name", "properties": { "name": "urn:ogc:def:crs:OGC:1.3:CRS84" } },
  "features": 
    [
      { 
       "type": "Feature", 
       "properties": { "id": 0 }, 
       "geometry": { "type": "MultiPolygon", "coordinates": 
         [ 
          [ 
           [
             [14.603147526122228,46.28089062404797],
             [14.620565004791132,46.28089062404797],
             [14.620565004791132,46.27187801679233],
             [14.603147526122228,46.27187801679233],
             [14.603147526122228,46.28089062404797]
            ] 
          ] 
          ] 
         } 
       }
      ]
    }
temporal_extent = ["2024-04-01", "2024-04-30"]
```

```python
public_url = "https://raw.githubusercontent.com/ESA-APEx/apex_algorithms/1fba3ba1bb02cc42c8758a60880b90adf8b54a10/algorithm_catalog/terravision/s2_sam/openeo_udp/s2_sam.json"
sam_cube = connection.datacube_from_process(
        "s2_sam", 
        namespace = public_url,
        temporal_extent = temporal_extent,
        spatial_extent= spatial_extent
        )
```

```python
job = sam_cube.create_job(
        title=f'S2 SAM processing for {temporal_extent[0]} to {temporal_extent[1]}')
job.start_and_wait()
```

```python
result = job.get_results()
result.download_files("sam_s2/")
```
