```python
import openeo

connection = openeo.connect("openeofed.dataspace.copernicus.eu").authenticate_oidc()
```

```python
spatial_extent = {
  "type": "Feature",
  "properties": {
    "ID": "598",
    "CNTR_ID": "ES"
  },
  "geometry": {
    "type": "Polygon",
    "coordinates": [[
      [-15.487987, 27.813076],
      [-15.489753, 27.814047],
      [-15.493856, 27.809741],
      [-15.492555, 27.808545],
      [-15.487809, 27.806311],
      [-15.485954, 27.809176],
      [-15.484348, 27.812444],
      [-15.486350, 27.812290],
      [-15.487987, 27.813076]
    ]]
  }
}
temporal_extent = ["2024-05-01", "2024-05-31"]
```

```python
public_url = "https://raw.githubusercontent.com/ESA-APEx/apex_algorithms/c0f8d549a866737a3c276f9838fa0a019e7975be/algorithm_catalog/terravision/s2_sam/openeo_udp/s2_sam.json"
sam_cube = connection.datacube_from_process(
        "s2_sam", 
        namespace = public_url,
        temporal_extent = temporal_extent,
        spatial_extent= spatial_extent
        )
```

```python
job = sam_cube.create_job(
        title=f'S2 SAM processing for {temporal_extent[0]} to {temporal_extent[1]}',
        out_format="NetCDF")
job.start_and_wait()
```
