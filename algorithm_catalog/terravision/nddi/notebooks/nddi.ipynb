```python
import openeo
connection = openeo.connect("openeofed.dataspace.copernicus.eu").authenticate_oidc()
```

```python
shape = {"west": 5.05, "south": 51.21, "east": 5.1, "north": 51.23}
start_date, end_date = ["2025-05-01", "2025-05-31"]
```

```python
public_url = "https://raw.githubusercontent.com/ESA-APEx/apex_algorithms/64eac11fde74e02089695c2bf4f3d4e4c67a1372/algorithm_catalog/terravision/nddi/openeo_udp/nddi.json"

nddi = connection.datacube_from_process(
    "nddi",
    namespace=public_url,
    temporal_extent=[start_date, end_date],
    spatial_extent=shape,
)
```

```python
job = nddi.create_job(
        title=f'NDDI calculation for {start_date} to {end_date}',
        out_format="NetCDF")
job.start_and_wait()
```

```python
results = job.get_results()
results.download_files()
```
